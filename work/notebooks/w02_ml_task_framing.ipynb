## 1. My lane as an ML task (type)

Binary classification is the best fit for this lane. The goal is to predict whether a content item is declining, using a yes/no label.

This is a decision-support task: if the model flags a page as likely declining, a content or SEO team can review or refresh it sooner.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd
from pathlib import Path

# Find the repo root by walking upward from the current working directory.
def find_repo_root(start: Path) -> Path:
    for path in [start, *start.parents]:
        candidate = path / "data" / "raw" / "content_refresh_anonymized.csv"
        if candidate.exists():
            return path
    raise FileNotFoundError("Could not find the FlyRank repository root containing the starter CSV.")

repo_root = find_repo_root(Path.cwd())
csv_path = repo_root / "data" / "raw" / "content_refresh_anonymized.csv"
df = pd.read_csv(csv_path)

# Create the derived binary target from the documented trend label.
df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)

# Show a small sample with relevant real columns and the derived target.
show_cols = [
    "content_id",
    "client_id",
    "content_type",
    "impressions_90d",
    "ctr",
    "avg_position",
    "trend_direction",
    "is_declining_label",
]

df[show_cols].head(10)
```

## 2. Target or proxy

## 2. Target or proxy

The target should be an observed label derived from the data, not a hand-made rule that we invent.

I will create a derived binary target called `is_declining_label`:

- `1` when `trend_direction == "down"`
- `0` for all other documented trend directions (`new`, `flat`, `up`, `stable`)

This is a better framing than trying to predict a vague concept directly from the raw text. The model would predict this label from other page features.

Important caution: `trend_direction`, `trend_pct`, `impressions_last_30d`, and `impressions_prev_30d` should not be used as model input features because they directly define the target. Also, `content_id` and `client_id` are identifiers, not useful model features.


## 3. Success metric

## 3. Success metric

A reasonable success metric is ROC-AUC.

In beginner-friendly terms, ROC-AUC measures how well the model ranks declining pages above non-declining pages. A higher value means the model is better at separating the two groups.

I would use ROC-AUC because this is a binary classification problem and the model is being used to help prioritize pages.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## 5. Why ML beats a fixed rule here

A simple fixed rule such as "refresh pages when traffic falls by 20%" is too rigid for this problem.

Real content performance is affected by many signals at once, including keyword difficulty, content type, freshness, traffic level, and engagement. A fixed rule can miss cases where several weaker signals together point to decline. A model can learn from many signals and make a more flexible decision.

The prediction would support a content or SEO team by helping them prioritize which pages deserve review or refresh.
